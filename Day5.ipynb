# Multiple Inheritance 

```python
class camera:
    def click_photo(self):
        print("Photo Clicked")
```

```python
class phone:
    def call(self):
        print("Calling .....")
```

```python
class smartphone(camera, phone):
    def phone(self):
        print( "I am smartphone ")
```

```python
s = smartphone()
```

```python
s.click_photo()
```

```python
s.call()
```

```python
s.phone()
```

# MultiLevel Inheritance 

```python
class animal:
    def eat(self):
        print("I am eating")
```

```python
class dog(animal):
    def bark (self):
        print("Barking")
```

```python
class puppy(dog):
    def weep(self):
        print("Weeping")
```

```python
tommy = puppy()
```

```python
tommy.bark()
```

```python
tommy.weep()
```

```python
class grandfather:
    def house (self):
        print("10 BHK apartment")
    
```

```python
class father(grandfather):
    def villa(self):
        print("Sea Facing Villa ")
```

```python
class child(father):
    def car (self):
        print("BMW")
```

```python
jack = child()
```

```python
jack.villa()
```

```python
jack.house()
```

```python
jack.car()
```

# hierarchical inheritance 

```python
class vehicle:
    def start (self):
        print("Vehicle started ")
```

```python
class car (vehicle):
    def wheels (self):
        print("4 Wheels ")
```

```python
class bike (vehicle):
    def wheels(self):
        print("2 Wheels ")
```

```python
c = car()
```

```python
c.start()
```

```python
c.wheels()
```

```python
b = bike()
```

```python
b.start()
```

```python
b.wheels()
```

# Hybrid Inheritance

```python
class vehicle:
    def start (self):
        print("Vehicle started ")
```

```python
class car(vehicle):
    def wheels (self):
        print("4 wheels ")
```

```python
class Battery:
    def charge (self):
        print("Charging ")
```

```python
class electric (car, Battery):
    pass
```

```python
e = electric()
```

```python
e.charge()
```

```python
e.start()
```
