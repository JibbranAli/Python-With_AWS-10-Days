```python
import boto3
```

```python
def list_buckets():
    s3_client = boto3.client('s3')
    
    response= s3_client.list_buckets()
    for bucket in response['Buckets']:
        print(bucket['Name'])
```

```python
list_buckets()
```

```python
def create_bucket():
    s3_client = boto3.client('s3', region_name="ap-south-1")
    bucket_name = input("Enter Unique bucket name : ")
    s3_client.create_bucket(Bucket=bucket_name, CreateBucketConfiguration = {
            "LocationConstraint": "ap-south-1"
        
    })
    print(f"Bucket :  {bucket_name} :    Created ")
```

```python
create_bucket()
```

```python
list_buckets()
```

```python
def upload_file_to_s3():
    list_buckets()
    print("--------------------------------")
    s3_client = boto3.client('s3', region_name="ap-south-1")

    bucket_name =  input("Enter Bucket name where you wanto upload the object :")
    
    local_file = input("Enter File name to upload : ")
    
    s3_file_name = local_file
    
    s3_client.upload_file(local_file, bucket_name , s3_file_name)

    print("!!!!!!!    FILE UPLOADED !!!!!!!!!")
```

```python
upload_file_to_s3()
```

```python
def download_file_from_s3():
    list_buckets()
    print("--------------------------------")
    s3_client = boto3.client('s3', region_name="ap-south-1")

    bucket_name =  input("Enter Bucket name where you wanto download the object :")
    
    s3_file_name =  input("Enter S3 Bucket Object Name to download ")

    download_path= f"downloads_{s3_file_name}"

    
    s3_client.download_file( bucket_name , s3_file_name, download_path)

    print("!!!!!!!    FILE DOWNLOADED !!!!!!!!!")
```

```python
download_file_from_s3()
```

```python
def list_all_objects():
    list_buckets()
    print("--------------------------------")
    s3_client = boto3.client('s3', region_name="ap-south-1")

    bucket_name =  input("Enter Bucket name where you wanto see ALL the object :")

    response = s3_client.list_objects_v2(Bucket=bucket_name)

    for obj in response['Contents'] :
        print(obj['Key'])
```

```python
list_all_objects()
```

```python
def delete_object():
    list_buckets()
    print("--------------------------------")
    s3_client = boto3.client('s3', region_name="ap-south-1")

    bucket_name =  input("Enter Bucket name where you wanto see ALL the object & Delete a Single  :")
    list_all_objects()
    file_name= input(f"Enter file name in {bucket_name} to DELETE :")

    s3_client.delete_object(Bucket=bucket_name, Key=file_name)

    print("File DELETED ", file_name)
    
```

```python
delete_object()
```
