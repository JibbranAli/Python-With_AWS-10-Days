```python
def launch_instance():
    """This is Used for launching the EC2 instance on AWS CLoud . Created by Jibbran ALi"""
    ec2_client = boto3.client('ec2', region_name='ap-south-1')
    try:
        response = ec2_client.run_instances( 
            ImageId = "ami-0b910d1016287a5e7",
            InstanceType = "t3.micro",
            MinCount= 1,
            MaxCount= 1,

            TagSpecifications=[
        {
            'ResourceType': 'instance',
            'Tags': [{'Key': 'Name', 'Value': 'Boto3',}]
            }
            ]
        )
        instance_id = response["Instances"][0]["InstanceId"]
        print(f"Instance Launched  : {instance_id}")
        return instance_id


    except ClientError as e:
        print(f"Failed to launch Instance : {e}")
        return None

        
```

```python
def get_instance_name(tags):
    if not tags:
        return "N/A"
    for tag in tags:
        if tag["Key"]  == "Name":
            return tag["Value"]
    return "N/A"
    
```

```python
def describe_all_instances():
    """This function will print table of All Instances in AWS Account"""
    ec2_client = boto3.client('ec2', region_name='ap-south-1')

    try : 
        response = ec2_client.describe_instances()

        header1 = "Instance ID"
        header2= "Name"
        header3 = "State"
        header4 = "Public IP"
        header5 = "Private IP"
        header6 = "Type"
        header7 = "AZ"
        print(f"{header1:<20}{header2:<20}{header3:<12}{header4:<16}{header5:<16}{header6:<12}{header7:<15}") 
        print("-" * 110)


        for reservation in response['Reservations']:
            for instance in reservation['Instances']:
                instance_id = instance['InstanceId']
                name = get_instance_name(instance.get('Tags'))
                state = instance['State']['Name']
                public_ip = instance.get('PublicIPAddress', "N/A")
                private_ip = instance.get('PrivateIPAddress', "N/A")
                instance_type = instance['InstanceType']
                az = instance['Placement']['AvailabilityZone']
                launch_time = instance['LaunchTime']

                print(f"{instance_id:<22}{name:<20}{state:<12}{public_ip:<16}{private_ip:<16}{instance_type:<12}{az:<15}")
                print(f"Launch Time : {launch_time}")
                print("_" * 110 )

    except ClientError as e:
        print(f"Failed to Describe Instance : {e}")
```

```python
def start_instance (instance_id):
    ec2_client = boto3.client('ec2', region_name='ap-south-1')


    try : 


        response = ec2_client.start_instances(InstanceIds=[instance_id])
        state = response['StartingInstances'][0]['CurrentState']['Name']

        print(f"Instance {instance_id} is now {state}")

    except ClientError as e:
        print(f"Could not start Instance {instance_id} : {e}")
```

```python
def stop_instances(instance_id):
    ec2_client = boto3.client('ec2', region_name='ap-south-1')
    try : 
        response = ec2_client.stop_instances(InstanceIds=[instance_id])
        state = response['StoppingInstances'][0]['CurrentState']['Name']

        print(f"Instance {instance_id} is now {state}")

    except ClientError as e:
        print(f"Could not Stop Instance {instance_id} : {e}")
        
```

```python
def reboot_instance(instance_id):
    ec2_client = boto3.client('ec2', region_name='ap-south-1')
    try :
        ec2_client.reboot_instances(InstanceIds=[instance_id])
        print(f"Reboot request Semt for instance {instance_id}")
    except ClientError as e:
        print(f"Could Not Reboot Instance {instance_id} : {e}")
```

```python
def terminate_instance(instance_id):
    ec2_client = boto3.client('ec2', region_name='ap-south-1')

    try :
        response=ec2_client.terminate_instances(InstanceIds=[instance_id])
        state = response['TerminatingInstances'][0]['CurrentState']['Name']
        print(f"Instance {instance_id} is now {state}")

    except ClientError as e:
        print(f"Could not Terminate  Instance {instance_id} : {e}")
```

## FILTERS 

```python
def list_running_instances():
    ec2_client = boto3.client('ec2', region_name='ap-south-1')
    try: 
        response = ec2_client.describe_instances(

             Filters=[{ 'Name': 'instance-state-name', 'Values': ['running']}]
        )

        for resrvation in response['Reservations']:
            for instance in resrvation['Instances']:
                print(instance['InstanceId'], "---", instance['State']["Name"])

    except ClientError as e :
        print(f"ERROR {e}")
```
