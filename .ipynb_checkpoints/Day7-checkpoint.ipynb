```python
import boto3
```

```python
def list_buckets():
    s3_client = boto3.client('s3')
    
    response= s3_client.list_buckets()
    for bucket in response['Buckets']:
        print(bucket['Name'])
```

```python
list_buckets()
```

```python
def create_bucket():
    s3_client = boto3.client('s3', region_name="ap-south-1")
    bucket_name = input("Enter Unique bucket name : ")
    s3_client.create_bucket(Bucket=bucket_name, CreateBucketConfiguration = {
            "LocationConstraint": "ap-south-1"
        
    })
    print(f"Bucket :  {bucket_name} :    Created ")
```

```python
create_bucket()
```

```python
list_buckets()
```

```python
def upload_file_to_s3():
    list_buckets()
    print("--------------------------------")
    s3_client = boto3.client('s3', region_name="ap-south-1")

    bucket_name =  input("Enter Bucket name where you wanto upload the object :")
    
    local_file = input("Enter File name to upload : ")
    
    s3_file_name = local_file
    
    s3_client.upload_file(local_file, bucket_name , s3_file_name)

    print("!!!!!!!    FILE UPLOADED !!!!!!!!!")
```

```python
upload_file_to_s3()
```
