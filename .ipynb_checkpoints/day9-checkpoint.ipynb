```python
import json
import boto3

bedrock = boto3.client("bedrock-runtime", region_name="us-east-1")

prompt = {
    "messages": [  # Fixed missing colon here
        {
            "role": "user",
            "content": [{"text": "What is serverless in cloud computing"}],
        }
    ]
}

response = bedrock.invoke_model(
    modelId="us.amazon.nova-lite-v1:0", body=json.dumps(prompt)
)

result = json.loads(response["body"].read())

print(result['output']["message"]["content"][0]["text"])
```
